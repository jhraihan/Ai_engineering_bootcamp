```python
import numpy as np
```

```python
a = np.array([2,3,4])
print(a.ndim)
print(a.shape)
print(a.size)
print(a.dtype)
```

```python
b = np.array([[2,3,4],[5,6,7]])
print(b.ndim)
print(b.shape)
print(b.size)
print(b.dtype)
```

```python
x = np.array([[2,3],[5,6],[7,8],[9,10]])
print(x.ndim)
print(x.shape)
print(x.size)
```

```python
a = np.zeros((2,2))
print(a)
print(a.dtype)
```

```python
b = np.ones((1,2))
print(b)
print(b.dtype)
```

```python
c = np.full((2,3), 7)
print(c)
print(c.dtype)
```

```python
d = np.eye(3)
print(d)
print(d.dtype)
```

```python
e = np.random.random((3,2))
print(e)
```

```python
f = np.arange(3)
print(f)
print(f.dtype)
```

```python
g = np.linspace(0, 10, 5)
print(g)
```
